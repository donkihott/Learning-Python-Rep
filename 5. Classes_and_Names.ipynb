```python
def scope_test():
    def do_local():
        spam = "local spam" # створює свою власну ізольовану змінну всередині себе. 
                            # Вона зникає одразу після виходу з функції. 
                            # Локальний spam у scope_test не змінюється.
    def do_nonlocal():
        nonlocal spam
        spam = "nonlocal spam"  # Локальний spam стає = "nonlocal spam"

    def do_global():
        global spam
        spam = "global spam"    # Python створює на глобальному рівні змінну spam = "global spam". 
                                # АЛЕ локальна змінна spam усередині scope_test() залишається недоторканою! 
                                # Вона все ще дорівнює "nonlocal spam".

    spam = "test spam"
    do_local()
    print("After local assignment:", spam) # test spam
    do_nonlocal()
    print("After nonlocal assignment:", spam) # nonlocal spam
    do_global()
    print("After global assignment:", spam) # nonlocal spam

scope_test()
print(spam) # global spam
```

```python
class Dog:

    kind = 'canine'         # class variable shared by all instances

    def __init__(self, name):
        self.name = name    # instance variable unique to each instance

d = Dog('Fido')
e = Dog('Buddy')
d.kind                  # shared by all dogs 
print(d.kind)   # - 'canine'
e.kind                  # shared by all dogs 
print(e.kind)   # - 'canine'
d.name                  # unique to d 
print(d.name)   # - 'Fido'
e.name                  # unique to e 
print(e.name)   # - 'Buddy'
print(d)    # <__main__.Dog object at 0x000001B89BF0FA10>
```

```python
class Dog:

    def __init__(self, name):
        self.name = name
        self.tricks = []    # creates a new empty list for each dog

    def add_trick(self, trick):
        self.tricks.append(trick)

d = Dog('Fido')
e = Dog('Buddy')
d.add_trick('roll over')
e.add_trick('play dead')

print(d.tricks)    # ['roll over']

print(e.tricks)    # ['play dead']
```

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.grades = []

    def add_exam(self, grade):
        """method, recording a new grade for him, as a float """
        self.grades.append(float(grade))

    def get_mean(self):
        """ method giving, For the Student, the average of its results."""
        if not self.grades:
            return 0.0
        return sum(self.grades) / len(self.grades)
    
    def __repr__(self):
        """ 🔑 Цей метод відповідає за красиве відображення студента у списках """
        return self.name

alice = Student('Alice')
bob = Student('Bob')
catherine = Student('Catherine')
daniel = Student('Daniel')

alice.add_exam(1)
alice.add_exam(2)
alice.add_exam(3)
print(alice.name, ':', alice.grades, 'Average of resalts -', alice.get_mean())

bob.add_exam(2)
bob.add_exam(3)
bob.add_exam(4)
print(bob.name, ':', bob.grades, 'Average of resalts -', bob.get_mean())

catherine.add_exam(3)
catherine.add_exam(4)
catherine.add_exam(5)
print(catherine.name, ':', catherine.grades, 'Average of resalts -', catherine.get_mean())

daniel.add_exam(4)
daniel.add_exam(5)
daniel.add_exam(6)
print(daniel.name, ':', daniel.grades, 'Average of resalts -', daniel.get_mean())
print()

class School:
    def __init__(self, name):
        self.name = name
        self.students = []

    def add_student(self, student):
        self.students.append(student)

    def get_mean(self):
        """ method giving, For the School, the average of the students averages. """
        if not self.students:
            return 0.0
        # Збираємо середні бали кожного студента та знаходимо їхнє середнє
        return sum(student.get_mean() for student in self.students) / len(self.students)
        
    def get_best_student(self):
        """ method, returning the best Student. """
        if not self.students:
            return None
        # Функція max знайде студента з найбільшим значенням, яке повертає get_mean
        return max(self.students, key=lambda student: student.get_mean())
    
    def __repr__(self):
        """ 🔑 Цей метод відповідає за красиве відображення школи у списках """
        return self.name

first = School('First')
first.add_student(alice)
first.add_student(bob)
print(first.name, ' students: ', first.students, 'Average -', first.get_mean(), 'The best is', first.get_best_student())

second = School('Second')
second.add_student(catherine)
second.add_student(daniel)
print(second.name, ' students: ', second.students, 'Average -', second.get_mean(), 'The best is', second.get_best_student())
print()

class City:
    def __init__(self, name):
        self.name = name
        self.schools = []

    def add_school(self, school):
        self.schools.append(school)
    
    def get_mean(self):
        """ method giving, For the City the average of the School averages."""
        if not self.schools:
            return 0.0
        return sum(school.get_mean() for school in self.schools) / len(self.schools)

    def get_best_school(self):
        """ method, returning respectively a School"""
        if not self.schools:
            return None
        # Функція max знайде школу з найбільшим значенням, яке повертає get_mean
        return max(self.schools, key=lambda school: school.get_mean())

    def get_best_student(self):
        """ method, returning respectively a Student"""
        if not self.schools:
            return None
        
        # 🔑 Збираємо найкращих студентів з усіх шкіл міста, ігноруючи порожні школи
        best_students_from_schools = []
        for school in self.schools:
            best_student = school.get_best_student()
            if best_student is not None:
                best_students_from_schools.append(best_student)
                
        # Якщо в жодній школі міста немає студентів
        if not best_students_from_schools:
            return None
            
        # Знаходимо абсолютного переможця серед найкращих студентів кожної школи
        return max(best_students_from_schools, key=lambda student: student.get_mean())
        
    def __repr__(self):
        """ Метод для красивого відображення міста у списках """
        return self.name

paris = City('Paris')
paris.add_school(first)
paris.add_school(second)
print('Schools of', paris.name, 'are:', paris.schools, 'Average is', paris.get_mean(), 'The best school is', 
      paris.get_best_school())
print('The BEST student in', paris.name, 'is', paris.get_best_student(), '!')
```
