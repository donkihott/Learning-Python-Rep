```python
import datetime

def friday_the_13th(start_date=None):
    """
    Find the next friday the 13th, starting from start_date.
    If start_date is not provided, the current date is used.
    """
    # If the date is not provided, use the current date
    if start_date is None:
        current_date = datetime.date.today()
    else:
        current_date = start_date
        
    # Step 1: Move to the nearest 13th date, if we are not already on it
    if current_date.day > 13:
        # Move to the next month
        if current_date.month == 12:
            current_date = datetime.date(current_date.year + 1, 1, 13)
        else:
            current_date = datetime.date(current_date.year, current_date.month + 1, 13)
    else:
        current_date = datetime.date(current_date.year, current_date.month, 13)

    # Step 2: Iterate through months until we find a Friday (weekday == 4)
    while True:
        if current_date.weekday() == 4:
            return current_date.strftime('%Y-%m-%d')
        
        # Move to the 13th date of the next month
        if current_date.month == 12:
            current_date = datetime.date(current_date.year + 1, 1, 13)
        else:
            current_date = datetime.date(current_date.year, current_date.month + 1, 13)

# --- ТЕСТУВАННЯ ---

# Тест 1: Перевірка для конкретної дати (13 лютого 2025)
test_date = datetime.date(2025, 2, 13)
print("Старт з 2025-02-13. Наступна п'ятниця 13-го:", friday_the_13th(test_date))
# Виведе: 2025-06-13 (бо лютий — це четвер, а червень — п'ятниця)

# Тест 2: Автоматичний пошук від сьогоднішнього дня
print("Старт від сьогодні. Наступна п'ятниця 13-го:", friday_the_13th())
```
