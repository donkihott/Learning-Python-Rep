### Conditional Operators and logical expressions

```python
a = int(input('Enter a:'))
b = int(input('Enter b:'))
c = int(input('Enter c:'))
if a > b and a > c:
    resalt = a*b*c
elif b > a and b > c:
    resalt = a+b+c
else:
    resalt = a-b+c
print('a =', a, 'b =', b, 'c =', c)
print(resalt)
```

```python
x = int(input("Enter x:"))
print(x)
if x > 7:
    f = x**3+7
    print('f for x>7:', f)
elif x < 7:
    f = 7*x-2
    print('f for x<7:', f)
else:
    f = 50
    print('x = 7, f =', f)
```
