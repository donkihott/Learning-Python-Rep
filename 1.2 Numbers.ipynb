Функція переведення чисел з однієї системи числення до іншої (цілі, беззнакові)

```python
### if number is whole and number > 0 ###
primarybase = 2
number = 11001011
newbase = 16
remainders = [] # list to store remainders
if primarybase > newbase:
    ### Function of successive division using the divmod function ###
    while number > 0:
        divisor = newbase # base of the number system
        number, remainder = divmod(number, divisor) # compute remainder and update number to the quotient
        remainders.append(remainder)   # store remainder
        if number == 0:
            print(f"remainders = {remainders}")
    remainders.reverse() # reverse the list to get correct order
    print(f" in base {divisor}: {remainders}")
else:   # if primarybase < newbase but NOT 16 base system!!!!!!!
    total_decimal = 0  # 1. Створюємо змінну для накопичення суми
    number_list = [int(d) for d in str(number)] # Create a list of integers
    number_list.reverse()
    print(number_list)
    while len(number_list) > 0:
        digit = number_list.pop()  # Get the last digit
        power = len(number_list)
        decimal_value = digit * (primarybase ** power)
        total_decimal += decimal_value # 2. Додаємо до підсумку
        if len(number_list) == 0:
            print(f"number {number} in base {newbase} is {total_decimal}")
```

```python
def dist():
    """
    Find the distance between the two furthest 
    apart values in a list
    1. user inputs a list of numbers, entered '=' to finish
    2. find the maximum and minimum values in the list
    3. return the difference between the maximum and minimum values
    """
    points = []
    while True:
        try:
            point = input("Enter a number, or '=' to finish: ").strip()
            if point == '=':
                break
            
            # Try converting to a number. First to float, 
            # to support both integers and fractional numbers.
            num = float(point)
            
            # If the number is an integer (e.g., 5.0), store it as an int for cleanliness
            if num.is_integer():
                points.append(int(num))
            else:
                points.append(num)
                
        except ValueError:
            print("Error: Please enter a NUMBER - not text!")

    # Check: whether the user entered any numbers before pressing '='
    if not points:
        print("Error: No numbers were entered!")
        return 0
    # Calculate the distance between the maximum and minimum values    
    return max(points) - min(points)

# Call the function and print the result
result = dist()
print("The distance is:", result)
```

```python
def from_roman_numeral(roman_numeral):
    roman_values = {
        'I': 1, 'V': 5, 'X': 10, 'L': 50, 'C': 100, 'D': 500, 'M': 1000
    }
    total = 0
    prev_value = 0
    for char in reversed(roman_numeral):
        value = roman_values.get(char, 0)
        if value < prev_value:
            total -= value
        else:
            total += value
        prev_value = value
    return total
```

```python
def fibonacci(n):
    """
    Generate the first n Fibonacci numbers.
    The Fibonacci sequence is defined as:
    F(0) = 0, F(1) = 1, and F(n) = F(n-1) + F(n-2) for n > 1.
    """
    if n < 0:
        return []
    elif n == 1:
        return [1]

    fib_sequence = [1, 1]
    for i in range(2, n):
        next_fib = fib_sequence[-1] + fib_sequence[-2]
        fib_sequence.append(next_fib)
    
    return fib_sequence
print(fibonacci(10))
```

```python
def how_to_pay(amount, currency):
    """- Programming_Basics-Python-Part-1.pdf"
    return a dict describing the easiest way to pay amount with the given currency.
    - amount is an amount to pay.
    - currency describe the currency as a list of existing coins or banknotes.
    how_to_pay(123, euro)        # To pay 123 euros:
    {100: 1, 20: 1, 2: 1, 1: 1}  # give 1 bill of 100€, one bill of 20€,
                                 # one coin of 2€, and one coin of 1€.
    - how_to_pay(100, [50, 20, 10, 5]) - the best way is to give:
    {50: 2}  # give 2 bills of 50
    """
    
    currency = euro if currency == "euro" else currency
    result = {}
    for coin in sorted(currency, reverse=True):
        count = amount // coin
        if count > 0:
            result[coin] = count
            amount -= coin * count
    return result
euro = [1, 2, 5, 10, 20, 50, 100, 200, 500]
print(how_to_pay(500, euro))
print(how_to_pay(123, euro))
print(how_to_pay(100, [50, 20, 10, 5]))
```

```python
def perfect_shuffle(deck):
    """
    Perform a perfect shuffle on a deck of cards.
    A perfect shuffle interleaves the two halves of the deck.
    For example, if the deck is [1, 2, 3, 4], the perfect shuffle would result in [1, 3, 2, 4].
    """
    deck = list(range(1, deck + 1)) if deck is int else deck
    mid = len(deck) // 2
    shuffled_deck = []
    for i in range(mid):
        shuffled_deck.append(deck[i])
        shuffled_deck.append(deck[i + mid])
    return shuffled_deck
#### test  if you shuffle 10 times a deck of 1024 cards, the deck returns to its initial state
deck = 1024
original_deck = list(range(1, deck + 1))
print(f"Original deck: {original_deck}")
shuffled_deck = perfect_shuffle(original_deck)
print(f"Shuffled deck: {shuffled_deck}")
while shuffled_deck != original_deck:
    shuffled_deck = perfect_shuffle(shuffled_deck)
    print(f"Shuffled deck: {shuffled_deck}")
```

```python
def check_adam_number(num):
    """Check if num is an Adam number"""
    num_str = str(num)
    num_reverse_str = num_str[::-1]
    num_reverse = int(num_reverse_str)
    print(f"Original number: {num}, Reversed number: {num_reverse}")
   
    square_of_orig = num ** 2
    square_of_reverse = num_reverse ** 2
    print(f"Square of original: {square_of_orig}, Square of reversed: {square_of_reverse}")

    square_of_reverse_str = str(square_of_reverse)
    print(f"Square of reversed as string: {square_of_reverse_str}")
    square_of_reverse_str_reversed = square_of_reverse_str[::-1]
    print(f"Reversed square of reversed as string: {square_of_reverse_str_reversed}")
    return square_of_orig == int(square_of_reverse_str_reversed)

print(check_adam_number(15))
```

```python
def mul(numbers):
    """
    Multiply all numbers in a list together.
    
    Arguments:
    numbers -- A list of numbers (int/float).
    
    Returns:
    The product of all numbers in the list.
    """
    product = 1
    for number in numbers:
        product *= number
    return product
print(mul([2, 3, 4]) + mul([1, 2]))
```
