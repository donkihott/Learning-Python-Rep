### Generator!!!

```python
def get_numbers():
    """Генератор, який віддає числа, поки не зустріне 0"""
    while True:
        try:
            num = int(input("Введіть натуральне число (0 для завершення): "))
            if num == 0: break      # the end of generation
            if num > 0: yield num   # generation
            if num < 0: print("Error: Enter number > 0!")
        except ValueError:
            print("Помилка: введіть ціле число!")

# Використовуємо max() з фільтрацією всередині
try:
    # default=None дозволяє уникнути помилки, якщо парних чисел не буде
    result = max((x for x in get_numbers() if x % 2 == 0), default=None)
    
    if result is not None:
        print(f"Найбільше парне число: {result}")
    else:
        print("Парних чисел не знайдено.")
except Exception as e:
    print(f"Сталася помилка: {e}")
```

Варінат 12:  
2.  Знайдіть суму елементів послідовності цілих чисел, що є кратними 
числу 5 і завершується числом 0.

```python
def get_whole_numbers():
    """ The generate sequence of whole numbers, the last - '0' """
    while True:
        try:
            number = int(input('Enter few whole numbers, the last - "0":'))
            if number == 0: break
            yield number
            print(number)    
        except ValueError:
            print("Error! Enter whole number")
print(get_whole_numbers.__doc__)
total_sum = 0
count = 0
for i in (num for num in get_whole_numbers() if num % 5 == 0 and num != 0):
    total_sum += i
    count += 1

if count > 0: 
    print(f'Sum of numbers multiplex number 5 = {total_sum}')
else: print('There are not multiplex number 5 numbers!')
```

```python
def get_natural_numbers():
    """ def get_natural_numbers()
    Generate sequence natural numbers:
    - user input few numbers
    - entering 0 finished sequence"""
    while True:
        try:
            number = int(input('Enter natural numbers (the last - 0):'))
            if number == 0: break
            if number < 0: print('Error! Enter number > 0!'); continue
            print(number)
            yield number 
        except ValueError:
            print('Enter numbers - NOT text')
print(get_natural_numbers.__doc__)
product = 1
count = 0
for x in get_natural_numbers():
    product *= x
    count += 1
if count > 0:
    geometry = product**(1 / count)
    print(f'Count of numbers: {count}')
    print(f'Product: {product}')
    print(f'Geometric mean = {geometry:.4f}')
else:
    print('No numbers were entered to calculate the geometric mean.')
```

```python
def real_numbers():
    """ Generate real numbers:
    (all numbers, including float and 0)"""
    while True:
        try:
            n = float(input('Enter any numbers - 0 is the last'))
            print(n)
            yield n
            if n == 0:
                break
        except ValueError: print('NOT string or symbols!!!')
biggest = float('-inf')
found = False
for i in real_numbers():
    if i > biggest:
        biggest = i
        found = True
if found:
    print(f'the maximum number is {biggest}')
else:
    print('There are not numbers except 0')
```
