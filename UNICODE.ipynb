```python
### Find all the emojis containing a heart
### Print characters all in a single line without any separator (not their names)
import unicodedata

# Iterate through the entire Unicode space
for n in range(0x0000, 0x10FFFF + 1):
    try:
        # Find for the word 'HEART' in the symbol name
        if 'HEART' in unicodedata.name(chr(n)):
            print(chr(n), end="")
    except ValueError:
        # Skip characters that do not have a name in Unicode
        continue
```
