```python
def is_a_dyck_word(word: str) -> bool:
    """
    Check if a given word is a Dyck word.
    
    A Dyck word is a balanced string of parentheses, where:
    A Dyck word is a word composed of matching opening and closing symbols.
    This implementation supports (), [], {}, <>, (((()))), ()(())(), AB, ABAB, ",.", dodo

    Returns:
    True if the word is a Dyck word, False otherwise.
    """
    matching_symbols = {
        '(': ')',
        '[': ']',
        '{': '}',
        '<': '>'
    }
    allowed_extra_chars = set(
        "./,|'\";:!@#$%^&*_+=?" +
        "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ"
    )
    stack = []
    
    for char in word:
        print(f'char is {char} ; matching_symbols[char] is {matching_symbols[char]}')
        # if char is key
        if char in matching_symbols:
            if matching_symbols[char] != char and matching_symbols[char] in word:
                stack.append(char)
                print(stack)
                if len(stack) == len(word):
                    return True



        elif char in allowed_extra_chars:
            pass
        else:
            return False

    # all opened symbols must be closed
    return not stack


print(is_a_dyck_word('()'))
```
