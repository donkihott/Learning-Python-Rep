1.  .  Підрахуйте  кількість  ком, лапок і крапок у рядку. 

```python
S = "Aston  approach  is  an  evolution  of  prior  conversations  about  'broad',\
     'narrow',  'strong'  and  'weak'  forms  of  AI."
# cells quotes and periods in a line
cells = S.count(",")
periods = S.count(".")
quotes = S.count("'")
print("The number of cells in the string is:", cells)
print("The number of periods in the string is:", periods)
print("The number of quotes in the string is:", quotes)
```

починаються на 'a' (велика або маленька буква).

Варіант 4: 
1.  Перевірте, чи є рядок паліндромом (розглядаючи 
тільки літери і цифри, без урахування регістру). 

```python
S = "A  common  example  of  a  neural  network  architecture  is  the  multilayer \
perceptron (MLP)."
# is the string a palindrome?
cleaned_string = ''.join(c.lower() for c in S if c.isalnum())
is_palindrome = cleaned_string == cleaned_string[::-1]
print(f"Is the string a palindrome? {is_palindrome}")
```

Варіант 4: 
2.  "This  includes  input  and  output  layers,  as  well  as  one  or  more  hidden 
layers.". Знайдіть і виведіть усі слова, що закінчуються на 's'.

```python
import string
s = "This  includes  input  and  output  layers,  as  well  as  one  or  more  hidden \
layers."
words = s.split()
for word in words:
    clean = word.strip(string.punctuation)
    if clean.lower().endswith('s'):
        print(clean)
```

Варіант 5: 
2.  . Перевірте, чи є рядок ASCII (В 
Python, рядок вважається ASCII, якщо кожний символ в рядку може 
бути представлений за допомогою ASCII кодування. ASCII кодування 
покриває символи в діапазоні від 0 до 127 включно).

```python
s = "In  statistics,  the  logistic  model  (or  logit  model)  is  used  to  model  the \
probability of a certain class or event."
# Is the string ASCII or Unicode?
if all(ord(char) < 128 for char in s):
    print("The string is ASCII.")
```

1.  . Видаліть всі прийменники. 

```python
s = "There is no abstract art. You must always start with something."
# to delete all prepositions
prepositions = ["is", "no", "with", "of", "the", "a", "an"]
words = s.split()
filtered_words = [word for word in words if word.lower() not in prepositions]
result = " ".join(filtered_words)
print(result)
```

```python
def is_anagram(left, right):
    left_letters = sorted(left)
    right_letters = sorted(right)
    return left_letters == right_letters
is_anagram("listen", "silent")
```
