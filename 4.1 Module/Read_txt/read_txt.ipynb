```python
### open a file named words.txt located in the same folder
### print the number of occurences of the letter e in its content
import pathlib
# 1. Read the contents of the file words.txt
letters_string = pathlib.Path("words.txt").read_text()
# 2. Count the total number of "e"
total_e = sum(1 for char in letters_string if char == 'e')
print(total_e)
```

```python
### open a file named words.txt located in the same folder
### compute the frequency of every ascii lowercase letters in its content
### display them according to this format: b: 0.02
import pathlib
import string

# 1. Read the contents of the file words.txt
letters_string = pathlib.Path("words.txt").read_text()
print(f'letters in file: {letters_string}')

# 2. Count the total number of lowercase ASCII letters in the file to accurately calculate the frequency.
total_ascii_lowercase = sum(1 for char in letters_string if char in string.ascii_lowercase)

# 3. Calculate and output the frequency for each letter present in the file.
past_letters = []
for letter in letters_string:   
    if letter in string.ascii_lowercase and letter not in past_letters:
        past_letters.append(letter)
        # Frequency = (count of this letter) / (total number of lowercase ASCII letters)
        # If the file is empty or contains no ASCII characters, we avoid division by zero.
        frequency = float(letters_string.count(letter)) / total_ascii_lowercase if total_ascii_lowercase > 0 else 0.00
        
        # Output in a format with two decimal places
        print(f'{letter}: {frequency:.2f}')
```
